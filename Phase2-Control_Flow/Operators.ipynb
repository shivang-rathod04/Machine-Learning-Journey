```python
# Comparison operators are used to compare the two values and return the result in form of boolean [either true or false] based on condition
```

```python
# Comparison operators are : <, >, <=, >=, !=, ==
# - syntax: value_1 comparison_operator value_2
# - these value can be anything : numbers, variables, functions, expressions
```

```python
# Eg:
# Values
print(2 < 3)

#Variables
x = 34
print(x > 45)

#functions
print(sum([1, 2]) < 34)

# Expressions:
print((24 - 12) >= 12)
```

```python
# Comparison in python for strings work in lexographical order.
# means under the hood, strings are numbers we call them, ASCII numbers
# This value determines which character is this
# Eg: 65 -> 'A'
# you can check this
```

```python
# what happens when you compare whole strings
print("Hello" < "India")
```

```python
# It checks the character from left side
# Only first character is checked 
# In this example : first character is 'H' and second character is 'I'
# ASCII of I is 73 and H is 72
# and 72 < 73 : that means True
```

```python
# You don't need to remember the ASCII, just follow H comes before I so I is bigger than H 
# follow alphabetical order
# Remember, python is case sensetive that means : uppercase and lowercase are different 
print('a' == 'A')
```

```python
# 2. Chained comparison
# - until now we saw python allows to compare two values using comparison operators 
# - but you can also compare more than two values, we call them as chained operations
# - most of the programing we take max 3 values
# - syntax: value_1 operator value_2 operator value_3
```

```python
# Check if age is between 18 to 30
age = 23
print(18 <= age <= 30)
```

```python
# This works as :
# It starts from left checks the left part : 18 <= age (23), as this is True so it returns --- True
# Then it checks the right side part : age <= 30 (23), as this is also True and it returns --- True
# True and True : True
# so the chained operation works like 18 <= age and age <= 30
# If both the expression/condition is True, then only it will return True
```

```python
# The most common mistake in python
# = and ==
# '=' means : making point an variable to an object inside memory [assining a value]
# '==' means : comparing the both objects values.
# If you use '=' instead of '==', python will give an syntax error
```
