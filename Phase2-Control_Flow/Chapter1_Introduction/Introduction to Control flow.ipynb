```python
# What is control flow ? 
# - It is a 'logic' that we write, that decides how our code will be executed.
# - Control flow usually includes three things : 
# - should i skip task ?
# - should i execute task ? 
# - should i repeat the task ? 
```

```python
# Earlier till now we have wrote 'linear coding'
# - meaning : 
# - from very start to end of your script. every line is executed.
# - that's not how real world programming works
# - Everything you see or perform actions is always in form of conditions. 
# - Eg: on crossroad taking right or left [depends on the condition that you create]
```

```python
# That's where control flow coding comes in
# - Control flow logic helps your program to able to think and decide based on the conditions [logic] that you create
# - control flow code including two things : 
# - 1. Conditionals
# - 2. Loops
```

```python
# Conditionals : 
# - This is desicion based programing, where particular part of your code is executed if it satisfy's required conditions.
# - this conditions are the questions [the logical part] that you create using "control flow expressions"
# - this expressions will be always evaluated in form of True or False [booleans]
```

```python
# loops : 
# - This is used when you want to perform the repatative task until the certain condition is matched
# - This is very much used in programing, think of like creating mini machine that can do your same work thousands of time until it staisfy condition 
```

```python
# Python allows to build a control-flow code using it's tools
# - This tools are divided into two segments
# - control flow statements
# - control flow expressions
```

```python
# 1. Control flow statments:
# - This are keywords that are used to create the control flow code
# - control flow statements are divided into two parts :
# - Conditionals : if, else, elif
# - loops : for loop and while loop
```

```python
# - in loops python gives 3 more keywords : 
# - break, continue and pass
# - this are the 'controls' for the control flow code that you write
# - it controls the behavior of your logic. 
# - this is just overview. this will be studied much later when we reach to loops
```

```python
# 2. Control flow expressions:
# - This is the 'logic' or questions that you build or create
# - this expressions result is fixed after evaluation: either it will return 'true' or 'false'
# - This expressions include : 
```

```python
# - values : True, False
# - functions: all(), any(), isinstance(), bool()
# - operators : logical operators {and or not} | membership operators {in, not in} | identity operators {is, not is} | comparison operators
# - this operators are the expressions that you create [logic part]
```

```python
# Control flow code in real world programming is used 95% of your code. 
# because it allows your static code to able to think and process based on the conditions [questions] that you create
# all of this will be studied in detail in phase 2 [control flow]
```
