```python
# any() and all() funcitons
```

```python
# we already studied booleans in phase 1 foundations
# where we saw that booleans are singleton objects inside python and it has two types 
# either true and false
# we also saw that anything inside python can be evaluated into booleans
# we also saw that in python few values always remain falsy [0, 0.0, None, 0j, "", [], {}, set(), (), set([]), (0), False]
```

```python
# but there are few control flow expression used with statments
# 1. any() : checks if any of the follwing object is truthy. if yes then returns true
# 2. all() : It's greedy, checks if all the following objects must be truthy, if any one is false. returns false
```

```python
# Eg: 
username = ""
phonenumber = "098-9909-99876"
email = ""

# suppose in form registration if the requirement is to fill any one information in-oder to register, and user enters one field or none can be evaluated 
# using any()
print(any([username, phonenumber, email]))

# But suppose all the fields are compulsary in-order to register
# using all()
print(all([username, phonenumber, email]))
```

```python
# Below is the few more examples How any() and all() can be used. [ignore the for loops. just see mechanics of how all and any works]
```

```python
# check if number is negative
numbers = [5, 8, 10, -3, 7]
print(any(value < 0 for value in numbers))
```

```python
# check if all numbers are positive
nums = [-2, -332, -4534, 212, -3]
print(all(value > 0 for value in nums))
```

```python
# Check whether any word starts with "A".
names = ["Raj", "Amit", "John", "Kiran"]
print(any(value[0] == 'A' for value in names))
```

```python
# Check whether all names have length greater than 3.
name = ["Rahul", "Amit", "John"]
print(all(len(value) > 3 for value in name))
```

```python
# Find whether the password list contains any empty password.
passwords = ["abc123", "xyz789", "", "pass456"]
print(all(passwords)) # means there exists an empty password inside list
```

```python
# Find if any word contains letter "a".
words = ["python", "java", "c++", "ruby"]
print(any((char == 'a' for char in word)for word in words))
```
