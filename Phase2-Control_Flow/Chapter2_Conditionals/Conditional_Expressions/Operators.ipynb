```python
# 2. Chained comparison
# - until now we saw python allows to compare two values using comparison operators 
# - but you can also compare more than two values, we call them as chained operations
# - most of the programing we take max 3 values
# - syntax: value_1 operator value_2 operator value_3
```

```python
# Check if age is between 18 to 30
age = 23
print(18 <= age <= 30)
```

```python
# This works as :
# It starts from left checks the left part : 18 <= age (23), as this is True so it returns --- True
# Then it checks the right side part : age <= 30 (23), as this is also True and it returns --- True
# True and True : True
# so the chained operation works like 18 <= age and age <= 30
# If both the expression/condition is True, then only it will return True
```

```python
# The most common mistake in python
# = and ==
# '=' means : making point an variable to an object inside memory [assining a value]
# '==' means : comparing the both objects values.
# If you use '=' instead of '==', python will give an syntax error
```

```python
# Logical operators in python are used to combine multiple boolean expressions
# This combination of multiple boolean expressions is your [logic]
# Or you can say the conditions that you create which are used with control flow statements
```

```python
# The logical operators include :
# - and 
# - or
# - not
```

```python
# 1. and
# - and is used when you want your all conditions to be true
# - if any one of the condition becomes false - the result will be false
```

```python
print(3 < 5 and 5 < 10) # True and True = True
print(3 < 5 and 5 > 10) # True and False = False
print(3 > 5 and 5 < 10) # False and True = False
print(3 > 5 and 5 > 10) # False and False = False
```

```python
# 2. Or:
# - any one of the condition is True then it will return True
```

```python
print(3 < 5 or 5 < 10) # True and True = True
print(3 < 5 or 5 > 10) # True and False = True
print(3 > 5 or 5 < 10) # False and True = True
print(3 > 5 or 5 > 10) # False and False = False
```

```python
# Python logical operators like and | or, are considerd to be short circuit operators.
```

```python
# for and : if left side is 'False', it won't check the right side and it will return the left part
# for or : if left side is 'True', it won't check the right side and it will return left part
```

```python
print('' and 90)
print(3 and {})
print(3 and {} and 90)
```

```python
print('' or 98)
print(34 or {})
print(34 or {} or 23)
```

```python
# You can also check hybrid short circuit
print('' and 23 or 45) # Returns : 45
print(34 and {} or 4) # Returns : 4
print('' or 34 and {}) # Returns : {}
print('' and {} or 23) # Returns : 23
print(34 or {} and 4) # Returns : 34
```

```python
# How the hybrid works is : 
# - 1. priority : in python and has higher priority then or 
# - 2. once it evaluates and it checks for or 
# Eg: 23 or {} and 34
# first it checks : {} and 34 - according to short circuit operations left is False so it won't check right side and it will return {}
# Then it will check : 24 and {}, and according to or left is True so it won't check right side and hence final output will be 
# 24
```

```python
# 3. not
# - 'not' inverses the following boolean result 
# - True -> False
# - False -> True
```

```python
print(not 43)
print(not '')
```

```python
# You can use multiple not's at once [Inverse of Inverse is always equal to orignal]

print(not not '')
print(not not 43)
```

```python
# We already studied about identity operators in phase1-foundations
# - we saw the concept of integer caching, usine None etc.
# - but let's see overview of identity operators
```

```python
# Identity operators are used to check if two objects are same inside memory or not
# - when we create object in python
# - we know, one of the property of object is the 'memory address'
# - so the memory of address of the object is same or different depends on mutablity and immuatablity of object
```

```python
# for immutablity
# - small integers are pre-created inside python
# - no matter how much objects you created in range of [-5 to 256]
# - all the variables will point to same object
```

```python
num1 = 23
num2 = 23
num3 = 23
print(num1 is num2 is num3)
```

```python
# This is concept of integer caching in python
# - that's why always use '==' to compare the object values. not is
# - This is implemetation details. means for different python version things are different
```

```python
# For strings
str1 = "Hellooo"
str2 = "Hellooo"

print(str1 is str2)
```

```python
# python is reusing same string object inside memory
# Instead of creating new one
# This is concept of string interning in pyhton
# during compilation process AST checks if strings are identical or not.
# Identical strings are re-used instead of creating new one. to save memory inside python
# Python pre-creates the dictonary for strings inside memory.
```

```python
# Again this is implementaion detail
```

```python
# For mutable objects
lst1 = [1, 2, 3]
lst2 = [1, 2, 3]

print(lst1 == lst2)
print(lst1 is lst2)
```

```python
# For mutable objects
# python creates an list object inside memory and this list object
# stores references of the elements inside memory
# these elements are seperately inside python
# and bounds variable lst1 to that list object
# lst1 -----> [List object]
#               |   |   |
#              1   2   3
# Then python creates another list object and this list object stores same references that list object 1 has
# we already know small integers are re-used. and int are immutables
# lst1 -----> [List object1]
#               |   |   |
#               1   2   3
#               |   |   |
# lst2 -----> [List object2]
```

```python
# That's why it returned false because there are two seperate list objects inside memory
# but references between them are shared
```

```python
print(lst1[0] is lst2[0])
```
