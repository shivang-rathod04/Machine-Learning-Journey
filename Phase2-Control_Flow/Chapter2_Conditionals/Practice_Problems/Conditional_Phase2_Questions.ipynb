```python
#Q1: Take three sides and check if they form a valid triangle. 
```

```python
a = int(input('Enter side 1: '))
b = int(input('Enter side 2: '))
c = int(input('Enter side 3: '))

if a < b + c and b < a + c and c < a + b:
    print('valid Triangle')
else:
    print('Invalid Triangle')
```

```python
#Q2 : If the sides form a valid triangle, determine whether it is equilateral, isosceles, or scalene.

x = int(input('Enter side 1: '))
y = int(input('Enter side 2: '))
z = int(input('Enter side 3: '))

if x < y + z and y < x + z and z < y + x:
    print(f'Valid Triangle')
    if x == y == z:
        print('Equilateral Triangle')
    elif x == y or y == z or z == x:
        print('Isosceles Triangle')
    else:
        print('Scalen Triangle')
else:
    print('Invalid Triangle')
```

```python
#Q4: Check if one of two given numbers is a multiple of the other.

n1 = 10
n2 = 20

if n1 == 0 or n2 == 0:
    print('number 0 cannot be multiple of anything')
elif n1 % n2 == 0:
    print('n1 is multiple of n2')
elif n2 % n1 == 0:
    print('n2 is multiple of n1')
else:
    print('Neither are multiple of each other')

# This logic : if you want both to be multiple of each other
if n2 >= n1:

    is_divisible = n2 // n1
    is_multiplicable = is_divisible * n1

    if is_multiplicable == n2:
        print('Both n1 and n2 are multiple of each other')
    else:
        print('They are not multiples of each other')
else:
    is_div = n1 // n2
    is_multiple = is_div * n2

    if is_multiple == n1:
        print('Both n1 and n2 are multiple of each other')
    else:
        print('They are not multiple of each other')
```

```python
#Q7: Take two numbers and determine whether both are even, both are odd, or one is even and one is odd.

num1 = 22
num2 = 27

if num1 == 0 or num2 == 0:
    print('number cannot be 0')
elif num1 % 2 == 0 and num2 % 2 == 0:
    print('Both numbers are even')
elif num1 % 2 != 0 and num2 % 2 != 0:
    print('Both numbers are odd')
elif num1 % 2 == 0 and num2 % 2 != 0:
    print('num1 is even and num2 is odd')
else:
    print('num1 is odd and num2 is even')
```

```python
#Q9: Take a day number (1–7) and print the corresponding day name. 

day = 7

match(day):
    case 1: print('Monday')
    case 2: print('Tuesday')
    case 3: print('Wednesday')
    case 4: print('Thursday')
    case 5: print('Friday')
    case 6: print('Saturday')
    case 7: print('Sunday')
    case default: print('invalid')
```

```python
#Q10: Take a month number (1–12) and print the number of days in that month (ignore leap years). 

month = 12

if month > 12 or month <= 0:
    print('Invalid month number')
elif month == 2:
    print('This month has 28 days')
elif month in (1, 3, 5, 7, 8, 10, 12):
    print('This month has 31 days')
else:
    print('This month has 30 days')
```

```python
#Q11: Write a program that asks the user for two numbers and prints Close if the numbers are within .001 of each other and Not close otherwise.

value1 = 23.356
value2 = 23.357

is_close = abs(value1 - value2)

if value1 == value2:
    print('Numbers are exact')
elif is_close <= 0.001:
    print('It is close')
else:
    print('It is not close')
```
