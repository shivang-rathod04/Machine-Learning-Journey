```python
#Q1: Take a character and check if it is a letter, a digit, or neither.

char = 'A'

if char.isalpha():
    print('It is a letter')
elif char.isdigit():
    print('It is a digit')
else:
    print('neither')
```

```python
#Q2: Take a number and print “Fizz” if divisible by 3, “Buzz” if divisible by 5, and “FizzBuzz” if divisible by both.

number = 15

if number % 3 == 0 and number % 5 == 0:
    print('Fizzbuzz')
elif number % 3 == 0:
    print('Fizz')
elif number % 5 == 0:
    print('Buzz')
else:
    print('neither')
```

```python
#Q3: Take three numbers and print the median value (neither maximum nor minimum).

lst = [1, 2, 3]
n = len(lst)

if n % 2 != 0:
    median = n // 2
    print(f'Median = {lst[median]}')
else:
    median = ((n // 2) + (((n + 1) // 2) + 1)) / 2
    print(f'Median = {median}')
```

```python
#Q4: Take 24-hour time (hours and minutes) and print whether it is AM or PM.

hour = 10


if hour < 0 or hour > 24:
    print('Invalid')
elif hour == 0 or hour == 24:
    print('AM')
elif hour >= 12:
    print('PM')
else:
    print('AM')
```

```python
#Q7: Take a single digit (0–9) and print its word form (“Zero” to “Nine”). 

value = 9

match value:
    case 0:
        print('Zero')
    case 1:
        print('One')
    case 2:
        print('Two')
    case 3:
        print('Three')
    case 4:
        print('Four')
    case 5:
        print('Five')
    case 6:
        print('Six')
    case 7:
        print('Seven')
    case 8:
        print('Eight')
    case 9:
        print('Nine')
    case default:
        print('Unknown')
        
```

```python
#Q8: Take a weekday number (1–7) and determine if it is a weekday or weekend. 

num = 3
if num < 0 or num > 7:
    print('invalid')
elif num in (1, 2, 3, 4, 5):
    print('Weekday')
else:
    print('Weekend')
```

```python
#Q9: Take electricity units consumed and calculate the bill as per slabs (using if-else). 

units = 260

if units < 0:
    print('invalid')
elif units >= 500:
    amount = units * 15
    print(f'Total bill : {amount}')
elif units > 300:
    amount = units * 10
    print(f'Total bill: {amount}')
elif units > 100:
    amount = units * 5
    print(f'Total bill: {amount}')
else:
    amount = units * 3
    print(f'Total bill: {amount}')
```

```python
#Q10: Take a password string and check basic rules (length ≥ 8 and contains at least one digit).

password = 'Shivang@123'
state = False
if len(password) >= 8:
    for char in password:
        if char.isdigit():
            state = True
            break
    if state:
        print('Valid Password')
    else:
        print('Invalid Password')
else:
    print('Password must contain atleast 8 characters')
```
