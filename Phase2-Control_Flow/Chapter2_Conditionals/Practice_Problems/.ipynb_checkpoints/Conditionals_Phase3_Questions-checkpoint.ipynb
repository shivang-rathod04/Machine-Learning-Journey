```python
#Q1: Take a 3-digit number and check if all digits are distinct. 
```

```python
number = 123
copy_of_number = number

# Exctracting the digits
d3 = copy_of_number % 10
d2 = (copy_of_number // 10) % 10
d1 = copy_of_number // 100

# Logic
if number < 100 or number > 999:
    print('Invalid number')
elif d1 != d2 and d2 != d3 and d3 != d1:
    print('All Three digits are distinct')
else:
    print('Not all digits are distinct')
```

```python
#Q3: Take a 4-digit number and check if the first and last digits are equal.

n = 1231

# Logic: 
if n < 1000 or n > 9999:
    print('Invalid input')
else:
    first = n // 1000
    last = n % 10

    # Main Logic
    if first == last:
        print('First and last digit are same')
    else:
        print('First and last digit are different')
```

```python
#Q4: Check whether a given integer is single-digit, double-digit, or multi-digit. 

value = 23

# logic
if value >= 100:
    print('Multi digit number')
elif value >= 10:
    print('Double digit number')
elif value >= 0:
    print('Single digit number')
else:
    print('Invalid number')
```

```python
#Q5: Check if a number is a multiple of 7 or ends with 7.

figure = 20

# logic:
if figure % 7 == 0:
    print('Number is multiple of 7')
elif figure % 10 == 7:
    print('Number ends with 7')
else:
    print('Neither of the condition is satisfied')
```

```python
#Q6: Take coordinates (x, y) and determine which quadrant the point lies in.

# co-ordinates
x = 3
y = 2

# logic
if x == y == 0:
    print('Origin')
elif x > 0 and y > 0:
    print('Ist Quadrant')
elif x < 0 and y > 0:
    print('IInd Quadrant')
elif x < 0 and y < 0:
    print('IIIrd Quadrant')
elif x > 0 and y < 0:
    print('IVth Quadrant')
else:
    print('Invalid Input')
```

```python
#Q7: Check if an amount can be evenly divided into 2000, 500, and 100 currency notes.
# For this problem logic is simple : just check for amount % 2000 == 0..but i have created broader version
amount = 90000

# logic

if amount >= 2000:
    # if amount is above 2000 then all currency note can be used
    if amount % 2000 == 0:
        print('amount can be evenly divided between 2000, 500 and 100')
    else:
        print('amount cannot be evenly divided between 2000, 500 and 100')
elif amount >= 500:
    # If amount is less than 2000 then only 500 and 100 note can be used
    if amount % 500 == 0:
        print('Amount is evenly disivisble between 500 and 100')
    else:
        print('Amount is not evenly divided between 500 and 100')
elif amount >= 100:
    # If amount is less than 500 then only 100 note can be used
    if amount % 100 == 0:
        print('Amount can be divided into 100')
    else:
        print('Amount cannot be divided into 100')
else:
    print('No currency can be used to divide the amount evenly')
```

```python
#Q9: Take two angles of a triangle and compute the third angle and check if it's triangle

angle1 = 90
angle2 = 40

angle3 = 180 - (angle1 + angle2)

print('It is a triangle' if angle1 + angle2 + angle3 == 180 else 'it is not a triangle') 
```

```python
#Q10: Check whether a number is a perfect square (without using the square root function). 

perfect_n = 25

is_perfect = int(perfect_n**0.5)

if is_perfect**2 == perfect_n:
    print(f'The {perfect_n} is perfect number')
else:
    print('It\'s not perfect number')
    
```
