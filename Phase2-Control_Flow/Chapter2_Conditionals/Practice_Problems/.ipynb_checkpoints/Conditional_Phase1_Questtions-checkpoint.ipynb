```python
# Q1: Take a number and print whether it’s positive, negative, or zero.
```

```python
number = 23

if number > 0:
    print('positive number')
elif number < 0:
    print('negative number')
else:
    print('zero')
```

```python
# Q2: Check if a number is even or odd. 

num = 12

if num % 2 == 0:
    print('Even number')
else:
    print('Odd Number')
```

```python
#Q3: Check if a number is divisible by 5

value = 35

if value % 5 == 0:
    print('Divisible by 5')
else:
    print('Not Divisible by 5')
```

```python
#Q4: Check if a number is divisible by both 3 and 5.

n = 15

if n % 3 == 0 and n % 5 == 0:
    print('Number is divisible by 3 and 5')
else:
    print('Number is not divisible by 3 and 5')
```

```python
#Q5: Check if a given year is a leap year.

year = 2000

if year % 4 == 0 and (year % 400 == 0 or year % 100 != 0):
    print('leap year')
else:
    print('not leap year')
```

```python
#Q6: Take two numbers and print the larger one. 

num1 = 30
num2 = 45

if num1 > num2:
    print(num1)
else:
    print(num2)
```

```python
#Q7: Take three numbers and print the largest. 

value1 = 32
value2 = 34
value3 = 12

if value1 > value2 and value1 > value3:
    print(value1)
elif value2 > value1 and value2 > value3:
    print(value2)
else:
    print(value3)
```

```python
#Q8: Take a temperature value and print “Cold”, “Warm”, or “Hot” using range conditions.

temp = 10

if temp >= 35:
    print('Hot')
elif temp >= 11:
    print('warm')
else:
    print('Cold')
```

```python
#Q9: Take a character and check if it’s a vowel or consonant.

char = 'f'

if char in ('a', 'e', 'i', 'o', 'u'):
    print('vowel')
else:
    print('consonent')
```

```python
#Q10: Take a character and check whether it’s uppercase, lowercase, a digit, or a special character

character = '2'

if 'A' <= character <= 'Z':
    print('Uppercase character')
elif 'a' <= character <= 'z':
    print('lowercase character')
elif '0' <= character <= '9':
    print('digit character')
else:
    print('Special characters')
```
