```python
#Q1: Take coordinates (x, y) and check if the point lies on the X-axis, Y-axis, or at the origin.

x = 1
y = 0

#logic
if x == y == 0:
    print('Origin')
elif x != 0 and y == 0:
    print('x-axis')
elif x == 0 and y != 0:
    print('y-axis')
else:
    print('neither')
```

```python
#Q2:  Take three numbers and check if they can form a Pythagorean triplet.

a = 7
b = 24
c = 25

if (c**2) == ((a**2) + (b**2)):
    print('Pythagorian Triplet')
else:
    print('Not Pythagorian Triplet')
```

```python
#Q4: Take time (hours and minutes) and print the smaller angle between the hour and minute hands
# 1 min : 6 degree [min hand]
# 12 hr : 360 degree and 1 hr : 30 degree [small hand]
# 1 min : 0.5 degree [small hand]
hour = 2
minute = 45
minute_angle = minute * 6
hour_angle = (hour * 30) + (minute * 0.5)

total_angle = abs(hour_angle - minute_angle)

if total_angle <= 180:
    print(total_angle)
else:
    small_hand = 360 - total_angle
    print(small_hand)
```

```python
#Q5: Take three numbers and check if they are in arithmetic progression. 
n1 = 1
n2 = 3
n3 = 5

if (n3 - n2) == (n2 - n1):
    print('AP')
else:
    print('Not AP')
    
```

```python
#Q6: Take three numbers and check if they are in geometric progression. 

value1 = 3
value2 = 9
value3 = 27

if (value3 // value2) == (value2 // value1):
    print('GP')
else:
    print('Not GP')
```

```python
#Q7: Take a 3-digit number and check if the sum of the first and last digit equals the middle digit. 

num = 264

n1 = num // 100
n2 = (num // 10) % 10
n3 = num % 10

if n1 + n3 == n2:
    print('Equal')
else:
    print('Unequal')
```

```python
#Q8: Take an integer (1–9999) and check if the sum of its digits is greater than the product of its digits.
num = 1142

if num < 1 or num > 9999:
    print('invalid')
elif num >= 1000:
    n1 = num // 1000
    n2 = (num // 100) % 10
    n3 = (num // 10) % 10
    n4 = num % 10

    if (n1 + n2 + n3 + n4) > (n1 * n2 * n3 * n4):
        print('Equal')
    else:
        print('Unequal')
elif num >= 100:

    n1 = num // 100
    n2 = (num // 10) % 10
    n3 = num % 10

    if (n1 + n2 + n3) > (n1 * n2 * n3):
        print('Equally')
    else:
        print('Unequal')
elif num >= 10:
    n1 = num // 10
    n2 = num % 10

    if sum(n1, n2) > n1 * n2:
        print('same')
    else:
        print('Different')
else:
    print('Single digit product and sum is same')
```

```python
total = 0
product = 1
count = 0
duplicate = num

while count < len(str(num)):

    n = duplicate % 10
    
    total += n
    product *= n

    duplicate = duplicate // 10
    count += 1

if total > product:
    print('Equal')
else:
    print('Unequal')
```
