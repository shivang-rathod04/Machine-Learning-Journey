```python
#understanding conditionals in python
```

```python
# Conditionals statments are also called decision based programming
# - where it allows to control the flow of your program
# - by executing the particular block of code based on some 'condition'
# - this condition act as a crossroad.
# - if it's True, it executes the statemetns inside the block else it skips it
```

```python
# Until now we did linear programming
# - where everyline of the code was executed
# - but control flow, allows to control your program and let it perform actions based on some condition.
# - This condition is created using boolean expressions and this is also called as logic
```

```python
# In decision based programming 
# - There are different forms of conditionals [branching]
# - it starts from basic to most advance
# - we will see them all one by one
```

```python
# Basic [if]
# - 'if' is the statement from which first condition starts
# - it works as
# - If following condition is True, then execute the block or skip the block entirely
```

```python
# Let's say we have grading system
# - we tell students above 90 must be graded to be 'A'
# - This works as :
# - programs starts
# -  reaches to condition : marks >= 90
# - if it's True : then student is graded as 'A'
# - if its False : skips the enire block
```

```python
# From this we understand that, if there's condition where we wan't to execute particular block only when it's true 
# we use if, This is also called "one-way decision"
```

```python
# There are some rules for if :
# 1. The first condition must always be the primary condition
# 2. If is must required, without if you can't use other statements as well
# 3. there must be only 1 if for each branch
# 4. The decion programing must start with if
```

```python
score = 80

if score >= 90:
    print('A')
```

```python
# else :
# - 'Two way decision' : when you have atleast two conditions.
# - else block is used when your all condition becomes false 
```

```python
# suppose you have condition where, you want to check if score is greater than 90, grade student as 'A' else grade it as 'F'
```

```python
if score >= 90:
    print('A')
else:
    print('F')
```

```python
# Rules for else:
# - It's not standalone, there must exist if to use else statement
# - There can't be more than 1 else in same chain
# - Else does not require any conditions, it's the statement used to indicate that if all condition becomes false, execute the else block
# - Else statement must be after the if 
# - It's not madatory to always add else [else is used when you have two way conditions]
```

```python
# Indentation in python
# - python is very strict with it's syntax
# - one of the concept is 'Indentation'
# - It means, an whitespace required before starting to write any statement
# - this whitespace is generally [2, 4, 8] but according to PEP - 8 standard, whitespace must be 4
# - The space indicates that line above it is part of the next line.
# - If there's no indentation, it returns syntax errors
```

```python
# elif 
# - multi - way decision 
# - elif is used when you have follow-up condition after the primary conditions
# - elif is only executed when the if condition fails
```

```python
# now the follow up question can be done in 3 ways
# - 1. if - 'if' condition fails, then it can move to 'elif' condition and execute it [3 way or multi way branching]
# - 2. using if inside if [called nested if]
# - 3. combining both the expressions together [logical operators]
```

```python
# Rules :
# 1. It's not standalone statment
# 2. There can be any number of elif in same chain
# 3. elif is always used after the if statment. and before else statement
```

```python
marks = 89

if marks >= 90:
    print('A')
elif marks >= 80:
    print('B')
else:
    print('F')
```

```python
# The above code says : 
# - if marks is greater than 90, execute the if part.
# - if, the primary condition false, go to the follow - up question and check if it satisfy the condtion
# - if it's True, execute the elif block and then terminate the control flow
# - else block is only executed when all the above conditions fails.
```

```python
# 1. Multiple elif structure [false side branching]
# - There can be more than one follow-up questions 
# - and python allows us to use multiple elif blocks branched sequentially
# - This is used when we have more than three follow-up conditions : like grading system, web-status codes, etc..
```

```python
# Eg: grading system
student_marks = 80

if student_marks >= 90:
    print('A')
elif student_marks >= 80:
    print('B')
elif student_marks >= 70:
    print('C')
elif student_marks >= 60:
    print('D')
else:
    print('F')
```

```python
# Python first checks the primary condition : is student marks [80] >= 90 ? False
# It moves to next follow-up question : is student marks [80] >= 80 ? True
# It executes the statement inside elif block and terminates the control flow. and gives output B
# This is used when you are dealing with multiple failures
```

```python
# In control flow the descion follows from top to bottom
# - as soon as condition is satisfied, python executes the block of code and exits the control flow
```

```python
# Nested if [True side branching]
# - until now we branched the Tree towards left side [false side]
# - we made a control flow where if condition becomes false, check the next follow up question
# - but we can branch the condition on right side [true side]
# - so structure becomes, if condition is True, check the another follow-up condition.
# - On each if statement, there is an else part for that branching
```

```python
# Eg: along with marks we want to check if student has submitted the project or not, if submitted grade is incremented with +
```

```python
result = 90
is_submitted = False

if result >= 90:
    if is_submitted:
        print('A+')
    else:
        print('A')
elif result >= 80:
    if is_submitted:
        print('B+')
    else:
        print('B')
elif result >= 70:
    if is_submitted:
        print('C+')
    else:
        print('C')
else:
    print('F')
    
```

```python
# How this works : 
# - It checks if the score of student is above [90] ? if the answer is True 
# - It checks the nested if, where it asks another questions has student submitted the project ? if True -> it executes the if block 'A+'
# - If false it will give 'A'
# - if the score is not above 90, it skips the primary condition and checks the next follow up question. is result >= 80, if true check another follow up question inside elif block
```

```python
# From this we saw we made a complex branching for true side and false side.
# remember : python always executes only one block in branching. that means there can be only one output
# The nested if is used when we have conditions where one depends on other. 
```

```python
# 3. Logical Operators:
# - Above we saw two ways of branching [false side [multiple elif] and true side [nested if]]
# - but logical operator allows us to combine the multiple boolean expressions into one.
# - this helps to reduce the complexity of branching and improves the readablity.
```

```python
student_score = 76
is_project_submitted = True

if student_score >= 90 and is_project_submitted:
    print('A+')
elif student_score >= 90 and not is_project_submitted:
    print('A')
elif student_score >= 80 and is_project_submitted:
    print('B+')
elif student_score >= 80 and not is_project_submitted:
    print('B')
elif student_score >= 70 and is_project_submitted:
    print('C+')
elif student_score >= 70 and not is_project_submitted:
    print('C')
else:
    print('F')
```

```python
# This is used when both condition are equally important to evalute
# for this example, there's redundacy, same logic again and again
# for this example we must use nested if. because one condition depends on other
```

```python
# Last topic : 
# - Independent if blocks
# - python allows to build independent if blocks
# - where one if block doesn't depend on other
# - their both task are different and executes one after other
```

```python
name = 'samyukthaa'
status = False

if name == 'shivang':
    print(f'Hello {name}')
else:
    print('You are not shivang, I can\'t help you to log in')

if status:
    print('Login successful')
else:
    print('Please login as shivang')
```

```python
# Q1: Validate the Quality and Correctness of Email Values
# Rules : Must not be empty -
# 2. must contain . and '@'
# 3. must contain exactly one '@' symbol
# 4. must end with '.com', '.org' or '.net
# 5. must not be longer than 254 characters
# 6. It must start and end with digit
```

```python
email = 'shivangrathod@gmail.net'.strip().lower()

if not email:
    print('Email must not be empty')
elif not ('.' in email and '@' in email):
    print('Email must exist "." and "@"')
elif email.count('@') != 1:
    print('There must be only one "@"')
elif not email.endswith(('org', '.net', '.com')):
    print('Incorrect Domain')
elif len(email) > 254:
    print('length of email is too long')
elif not(email[0].isalnum() and email[-1].isalnum()):
    print('Email must start and with letter or digit')
else:
    print('valid email')
```

```python
# Password validation
password = 'Shivang1'
email = 'Shivang1'
if not password:
    print('Password cannot be empty')
elif len(password) < 8:
    print('length of password must atleast be 8')
elif password.isupper():
    print('Password must include atleast 1 lowercase')
elif password.islower():
    print('Password must include atleast 1 uppercase')
elif password == email:
    print('Password must not be equal to email')
elif ' ' in password:
    print('Password must not contain any spaces')
elif not(password[0].isalnum() and password[-1].isalnum()):
    print('Password must start and end with letter or digit')
else:
    print('password is valid')
```

```python
# Ternary if else
# Ternary if - else is an short hand property of writing two way branching structure
# - it reduces line of code and improves readablity
# syntax: true if condition else false  
```

```python
# To check if age is above 18 or not
age = 23
print(True if age >= 18 else False)
```

```python
# You cannot use elif structure in ternary-if else
# But you can use independent if else structure when you have more than one condition
score = 89
print('Pass' if score >= 90 else 'Good score' if score >= 80 else 'bad')
```

```python
# Ternary if-else is used when you are dealing with simple if-else structure [two way branching or one-way branching]
# For complex branching : use multi-way branching techniques [nested, multiple elifs, logical operator]
```

```python
#Last concept
# Match case
# match case is used to compare a value with multiple values

# let suppose you want to check if user hits 2 - then it must print Tuesday, if 1 then monday
# This can be done using if-elif-else strcuture but it becomes complex to read and write and reuse same variable again and againb
# That's where match case makes things easy
```

```python
day = 3

match day:
    case 1:
        print('Monday')
    case 2:
        print('Tuesday')
    case 3:
        print('Wednesday')
    case 4:
        print('Thursday')
    case 5:
        print('Friday')
    case 6:
        print('Saturday')
    case 7:
        print('Sunday')
    case default:
        print('Unkown')
```

```python
# Match case is only used to compare the values. it cannot be used for complex logic, or dealing with complex expressions
```
