```python
# Understanding while loops in python
```

```python
# While loops are another catagory for automating the task
# - unlike for loop, while loop keeps Iterating until the condition is True.
# - when condition becomes false, It jumps out from the loop and Terminates
```

```python
# - what makes while loop different from for loop
# - The for loop iterates on the sequence of items.
# - where start and end are already defined
# - we iterate on each item and perform the manuplation on each one of them
```

```python
# But in while loop, 
# - there is no sequence. meaning
# - while loop is used when we don't know our start
# - we don't know how many number of iterations we need to do
# - all we know is just the condition where iteration stops.
```

```python
# In simple terms : 
# - while loop is used when number of iterations is not known
# - for loop when performing the manuplation on sequence of items.
```

```python
# Syntax:
# - we use while keyword
# - while stopping_condition:
#  -     action
```

```python
# There two ways of writing the while loop
# - while with condition
# - while with True
```

```python
# 1. while with condition
# - this is used when we know, where to stop our automation of loop i.e knowing the stopping condition
# - this is helpful when we are dealing with extraction, counting, summing, manuplating etc..
```

```python
# There are three things to understand while loop
# - 1. Inititalization [starting count for our state variable]
# - 2. stopping condition [condition which will eventually become false and will jump out of loop]
# - 3. Updation [Updating the state variable]
```

```python
# - If any one of the thing fails, It will cause problems in while loops
# - and there are only three possiblites:
# - Infinite loop : This occurs when you provided wrong stopping condition or not updating your state variable
# - loop never executes : if your initilazation fails the stopping condition before even loop has started
# - wrong output : This occurs when you are applying wrong logic even if the all three states are satisfied
```

```python
count = 0 # Initilization

while count <= 10 : # Stopping Condition
    print(count)
    count += 1 # Updation
```

```python
# Note: 1. While loop is always dependent on a state variable.
# - The state variable decides how the while loop will behave 
```

```python
# 2. Never use unnessarry variable as a counter. It will take more space inside memory 

base = 2
power = 5 # used the power as counter instead of using another variable
result = 1
while power >= 1:
    result *= base
    power -= 1
print(result)
```

```python
# always check if you require the particular value after the manuplation ?
# - if no, use that as your stopping condition/ state variable
```

```python
# 2. While with True
# - This is used when you don't know stopping condition, or you don't know how many number of iteration are required
# - Unlike while with condition, this is already in state of infinite loop
# - and there is only one way to exit from the loop
# - add one extra condition that and break the loop.
# - This is used when you are dealing with task like 
# - [password validation]: Because you don't know how many times user will give correct password
# - Taking form input
# - server and requests
# - api calls etc..
```

```python
correct = '123'

attempt = 0
while attempt < 3: # no more attempt then 3
    password = input('Enter your pin')
    if password == correct: # Only accepts the correct password, if inncorrect It keeps asking the password
        print('Access Granted')
        break
    print('Invalid Password, Please Try again')
    attempt += 1
else:
    print('You are Blocked, Please wait for 40 seconds')
```

```python
# While loop with flag variables
is_running = True
while is_running:
    ask = input('Do you want to quit (yes/no): ')
    if ask == 'yes':
        is_running = False
```
