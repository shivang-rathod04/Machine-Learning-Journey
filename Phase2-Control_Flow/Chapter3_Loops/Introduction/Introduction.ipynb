```python
# loops are used to control the flow of our program
# In simple terms,
# Loops are used to repeat the same task again and again in order to do 'something'
# The 'something' is nothing but the logical part or the task we want to do
```

```python
# Up until now we saw
# Linear programming : where every line executes sequentially from top to end
# Control flow programming : decision based : where program executes based on condition
```

```python
# But still our program is hardcoded.
# It's able to make decisions but same task is need to be done twice, or thrice or even N times
# That's where the problem arises
# We don't want to write same logic 1000's of time for each block of code
# We need generalised, scalable solution where same logic can be applied to all the cases or conditions for n number of times
# This is where dynamic programming starts
# That's where we used iterative staments [loops]
```

```python
# We attach our logic to the loops and tell it to repeat the same logic to each item N number of times.
# Until certain condition is matched
```

```python
# Therefore,
# Python provides Tools to create a Iterative based programs
```

```python
# 1. For loop
# 2. While loop
```

```python
# Other with these tools we get controls to control the execution of the loops
# Keywords : break, continue and pass
```

```python
# In this entire chapter we will be understanding loops and it's controls in detail
# along with short hand property called : Comprehensions
# and techniques like : zip() and enumarate()
```

```python
# Rule to keep in mind
# Loops are not just treated as block of code or just a syntax
# They must be treated as scalable Transformations [dynamic solution for real world problem]
```

```python
# 1. Up until now you hardcoded for 3-4 conditions or numbers
# But from now the ultimate goal must be to build a generalised dynamic solution for N number of tasks
# 2. There's concept of time complexity and space complexity, for reading N number of items in loop it takes N number of time [means it's O(N)]
# Your ultimate solution must be always in O(n), if you can find a mathamatical formula to do same task that bypass the loop. Then that's elite level
# Aim always for O(1) as it's constant 
```

```python
# 3. Always trace your logic and state first on paper before coding
# Because at the end loops just change the state of variables again and again 
```
