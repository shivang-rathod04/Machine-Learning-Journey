```python
# Understanding Nested for Loops in python
```

```python
# In python, we learnt in conditionals that when one condition depends on other we use a nested if blocks
# - similarly in for loops we can nest them one inside another
# - but their behavior is different.
```

```python
# A simple for loop:
n = 5
for i in range(n):
    print(i, end = ' ')
```

```python
# For a single for loop, The time complexity is always O(n)
# - means for n inputs, number of work done by loop is also n times
# - That makes a loop efficient.
```

```python
# Nested for loops:
# - These are the loops where one loop runs inside another one
# - It's written as :
# - for ________:
# -     for ______:
```

```python
# To understand nested for loop better, you need to understand that there are two types of nested for loops
```

```python
# 1. Independent nested for loops
# 2. Dependent nested for loops
```

```python
# 1. Independent nested for loops
# - This is when inner loop dosen't depend on outer loop
# - Eg:
```

```python
# Independent
for i in range(3):
    for j in range(4):
        pass
```

```python
# You can see that range of outer loop is 3 and range of inner loop is 4
# - that means inner loop is independent of outer loop
# - It will become depedent when range of j depends on i
# - Eg:
```

```python
# Dependent
for i in range(3):
    for j in range(i):
        pass
```

```python
# Now let's understand each catagory in detail
```

```python
# 1. how independent nested for loops works
# - so when we write
# for i in range(3):
#    for j in range(4):
# then, for each i item, the inner loop j runs 4 times
# that means for each cycle of one outer loop, the inner loop executes n times
```

```python
# Eg:
for i in range(3):
    for j in range(4):
        print(f'({i},{j})', end = ' ')
```

```python
# You can see that for each item of outer loop 
# - inner loop ran for times (0,0), (0,1), (0,2), (0,3), ....so on
```

```python
# Therfore for n times :
# outerloop : n and innerloop: n
# Time complexity : O(n^2)
```

```python
# Similarly you can use more two nested loops
```

```python
# Eg:
for i in range(3):
    for j in range(4):
        for k in range(5):
            print(f'({i},{j},{k})', end = ' ')
```

```python
# for three nested independent loops concept remain same
# - for each item of i, the inner loop j run n times, and for each item of j, deep inner loop run n times
# - so when,
# - python creates a iterator object inside memory
# - The outer loops asks the next() item from iterator [0-2]
# - Iterator returns the value 0 and bounds the loop variable i -> 0
# - Then it goes inside and inner loop j asks the next() item from iterator object [0-3] and it returns 0
# - Then it bounds the item 0 -> j [so far i -> 0 and j -> 0]
# - then the deep inner loop asks the next item from the iterator object [0-4]
# - iterator object returns the value 0 and bounds the loop_variable k - to that item
# - Therefore it prints (0, 0, 0)
```

```python
# - now the deep loop variable asks the next() item from iterator object and state of k -> 0 to 1
# - and prints(0, 0, 1)
# - the cycle continues of deep inner loop till last item [4]: (0,0,4)
# - as the last item is over, it terminates the loop and goes back to the 2nd inner loop
# - the second inner loop runs and now it asks next item. state of variable j changes from 0 -> 1
#- and it starts the cycle for deep inner loop again. (0, 1, n) where [n = 0 to 4)
```

```python
# for three nested loops 
# Time complexity - n^3
```

```python
# 2. Dependent nested loops
# - dependent nested loops are those where inner loops depend on outer loop variable
# - and growth of inner loop depends on the current state outer loop variable
# - for each growth it runs n times for each item of outer loop i
```

```python
# that is : 
# for 0 : Outer loop, inner loop : 0 [nothing]
# for 1 : outer loop, inner loop : 1
# for 2: outer loop, inner loop : 2 [0, 1]
# for 3: outer loop, inner loop : 3 [0, 1, 2]
# so on...
```

```python
for i in range(3):
    for j in range(i):
        print(f'({i},{j})', end = ' ')
```

```python
# How it executes:
# - python creates a iterator object_1 inside memory of range 0 - 2
# - loop asks the next() item from the iterator and iterator returns the value 0
# - this value 0 is bound to the loop_variable i [i.e i = 0]
# - it then goes inside inner loop and python sees a range(i) and i = 0
# - therefor range of 0 is nothing
# - The inner loop terminates and goes back to the start [outer loop]
# - this time outer loop asks the next item from iterator object_1 and iterator returns the value 1
# - therfore state of outer loop variable changes from 0 -> 1
# - then it goes inside the inner loop and python creates an iterator object_2 of range(i) where i = 1 [range(1)]
# - the inner loop executes one time 
# - therefore, it prints (1, 0)
# - the inner loop terminates and goes back to start, and the outer loop demands next() item 2
# - iterator returns value 2 and state of outer loop variable changes from 1-> 2
# - then it goes inside inner loop and this time range(i) is range(2), that means 0 , 1
# - for item 2 the inner loop will run two times 0, 1
# - therfore following output will be 
# - (1, 0), (2, 0), (2, 1)
```

```python
# Thefore time complexity 
# - outerloop runs (n) times
# - inner loop runs (n - 1) times
# n (n -1) / 2
# - time complexity is O(n^2)
```

```python
# The time complexity is same
# - but their behaviour is different
# - one runs both n times independently and other runs dependently with n and n - 1 [summition]
```

```python
# for 3 dependent loops - patterns
for i in range(2):
    for j in range(2):
        for k in range(j):
            print(f'({i},{j},{k})', end = ' ')
```

```python
# Outer loop creates an Iterator object
# - Iterator object ranges from [0 - 1]
# - when i = 0, j = 0
# - k is depend on j, thefore range of k -> 0 [nothing] , loop terminates and moves back to j
# - when j = 1, range of k = 1 [0]
# - therfore i = 0, j = 1 and k = 0 [prints(0, 1, 0)]
# - it moves back to outer loop, state of i  changes to 1
# - when i = 1, j = 0, k -> Nothing [range of 0 is nothing]
# - when i = 1, j = 1, k = 0 [range of 1 -> 0]
# - it prints (1, 1, 0)
```

```python
for i in range(2):
    for j in range(i):
        for k in range(2):
            print(f'({i},{j},{k})', end = ' ')
```

```python
# Outer loop creates an Iterator object
# - Iterator object ranges from [0 - 1]
# - when i = 0, j -> depends on the outerLoop variable 
# - Thefore it executes nothing. it moves back to start and this time i = 1
# - when i = 1, range of j = 1 [0]
# - for 0 - k executes 2 times - as k is independent 
# - (1, 0, 0), (1, 0, 1)
```

```python
for i in range(2):
    for j in range(2):
        for k in range(i):
            print(f'({i},{j},{k})', end = ' ')
```

```python
# Outer loop creates an Iterator object
# - Iterator object ranges from [0 - 1]
# - when i = 0, j -> 0, k -> depends on i 
# - Thefore k executes nothing. it moves back to start of j and this time j = 1
# - It still executes nothing as k = 0 thefore the loop terminates and it moves back to outer loop again
# - for i = 1, j = 0 and range of k = 1 [0] 
# - it executes and moves back to j
# - it then j -> 1 and k -> 0 and prints
# - (1, 0, 0), (1, 1, 0)
```

```python
for i in range(3):
    for j in range(i):
        for k in range(j):
            print(f'({i},{j},{k})', end = ' ')
```

```python
# Outer loop creates an Iterator object
# - Iterator object ranges from [0 - 2]
# - when i = 0, range of j = 0 [that means nothing] loops back to start
# - when i = 1, range of j = 1, range of k = 0 [that means nothing] and loops back to start
# - when i = 2, range of j = 0, 1
# - for each item of j, k run n times
# - for j = 0, k [nothing executes]
# - for j = 1, k = 0, i = 2
# - it then j -> 1 and k -> 0 and prints
# - (2, 1, 0)
```
