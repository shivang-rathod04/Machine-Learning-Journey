```python
# Understanding for-else in python
```

```python
# Up until now we saw for loops
# - what they are, how it executes and controls in loops
# - but python allows us to use a else statement with a for loop
```

```python
# We saw in if-else, where if ceratain conditions are false at last else block executes
# - but in for loop case is different
```

```python
# The else block only executes if for loop is successfully completed
# - means no abrutptly terminating the loop
# - If loop is terminated abruptly the else block never executes
```

```python
# Syntax: 
# - for loop_variable in sequence:
# -     action
# - else:
# -     action
```

```python
# Eg:
for i in range(3):
    print(i)
else:
    print('Loop Terminated')
```

```python
# But using else just after loop terminates successfully doesn't make any sense
# - you can use same output without else

for i in range(3):
    print(i)
print('Loop Terminated')
```

```python
# Then where for-else loop is used ?
# - this is used when you are willing to get result with a report or you can say
# - when you want proper validation to see if loop was executed properly or not
# - to do that we use else with a break keyword
```

```python
for i in range(11):
    if i % 2 == 0:
        print(f'Even number are found')
        break
    print(i)
else:
    print('All the numbers are odd')
```

```python
# So, we know else only executes when loop is terminated successfully 
```

```python
#Q: check if file_list doesn't contain any duplicates, if they consist then exit immediatly

file_list = ['data.txt',
            'names.csv',
            'report.xlsx',
            'data.txt',
            'report.csv']

for file in file_list:
    if file_list.count(file) > 1:
        print(f'{file} Duplicate was Found')
        break
else:
    print('No duplicates exists in a file_list')
```
