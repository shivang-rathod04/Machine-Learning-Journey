```python
# Understanding controls in loops
```

```python
# Up until now we saw how we can automate our task
# but we were executing the sequence from start to end
# There was no control for loops [ just like cars without break]
```

```python
# so python provided three controls for our loops
# 1. break
# 2. Continue
# 3. Pass
```

```python
# Each of them serve different purpose
```

```python
# 1. break
# - break is used when you want to abruptly terminate the automation for task
# - you can use break anywhere in the loop [with or without condition]
# - but when loop breaks, it abruptly aries it's special error [stopIteration] and jumps out from the loop
# - This is used when you : 1. when you want to find a specific value, 2. avoid infinte search and etc
```

```python
for i in range(1, 6):
    if i == 3:
        print('Unkown value found')
        break
    print(i)
```

```python
# How it executes is :
# - the loop asks the next() item from the sequence [1- 5]
# - the iterator returns the value 1 and assigns it to loop variable [i] i.e i = 1
# - it then checks the condition [is i == 3 i.e 1 == 3] which is false and continues to print (1)
# - then it starts again by asking next() item from sequence
# - this time iterator returns value 2 and state of loop variable changes from 1 -> 2
# - then inside the loop it checks [is 2 == 3] that is false
# - and then executes the printt(2)
# - again it reaches to the top and asks next() item from the sequence
# - and the iterator returns the value 3 and state of loop variable changes from 2 -> 3
# - therfore now it checks if 3 == 3 : that is true and executes statement inside if block
# - first it prints 'Unkown value found' and then looks at break keyword
# - this indicates loop to break [jump out] abruptly and terminate the automation of task
# - therefore it prints -> 1 , 2, Unkown value found
```

```python
# Eg:
names = ['Rahul', 'Soumya', ' ', 'Kumar', '    ']

for words in names:
    if not words.strip():
        print('Unknown value found')
        break
    print(words)
```

```python
# Continue
# - Continue is used to skip one whole loop cycle
# - This is different from break
# - in break, we terminate the loop abruptly if we find false value 
# - whears in continue we tell loop, 'Okay, if you find the particular value skip it'
```

```python
# Why we need contiue is because - in break termination ends isntantly. It doesn't care about remaining values
# whears loop traverse through each item in continue but skips the loop cycles when it caches the required condition
```

```python
for value in range(1, 4):
    if value == 3:
        print('Unwanted values')
        continue
    print(value)
```

```python
# How it works : 
# - Python creates an iterator object inside memory of range 1 to 4 - 1 : 1 to 3
# - loop asks the next item from iterator
# - iterator gives the value 1 and bounds the value 1 to the loop varaible 'value'
# - then it checks if value == 3 [1 == 3] that is false and continues to print 1
# - Then loop again asks the next() item from iterator
# - iterator gives value 2 and state of loop variable changes from 1 -> 2
# - Then it checks if value [2] == 3 : that is false
# - and continues to print(2)
# - Then loop again asks the next() item from the iterator object
# - iterator object gives value 3 and state of loop variable changes from 2 -> 3
# - then it checks if 3 == 3 : that is true and executes the statement inside if
# - it prints('Unkown values') and then see's key word continue
# - and skips the entire loop cycle. for this case 3 was the last item. the last check for next loop is not done. it terminates 
```

```python
# Eg:
for word in names:
    if not word.strip():
        print('Empty value found')
        continue
    print(word)
```

```python
# pass
# - pass is used as empty placeholder
# - it is described as do nothing
# - this is used when you don't know what next operation you have to do but still you want to keep logic intact
# - you use pass
# - by using it loop sees it as to ignore the particular statement and keep the execution flow intact
```

```python
for i in range(1, 4):
    if i == 2:
        pass # multiply the value by 6
    print(i)
```

```python
#Eg:
for word in names:
    if not word.strip():
        #word = word.strip().replace('', 'Null')
        pass # Replace empty values with Null
    print(word)
```

```python
# Q1: Loop through list of days, and print only working days and skip weekends

week = set(['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])

for day in week:
    if day in ('Saturday', 'Sunday'):
        continue
    print(f'Weekdays = {day}')
```

```python
emails = set(['data@gmail.com',
          'barra@yahoo.de',
          'Drop Table Users',
          'maria@gmail.com'])

for email in emails:
    if '@' not in email.strip():
        break
    print(email)
```
