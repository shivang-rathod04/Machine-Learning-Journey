```python
#Q1: Count Digits
n = 1239
count = 0

while n > 0:
    n = n // 10
    count += 1
print(count)
```

```python
#Q2: Reverse a Number
x = 12321
duplicate = x
reverse = 0

while x > 0:
    digit = x % 10
    reverse = reverse * 10 + digit
    x //= 10
print(reverse)
```

```python
if reverse == duplicate:
    print(True)
else:
    print(False)
```

```python
#Q3: Find all divisors
number = 12
divisors = []
n = number // 2

for i in range(1, n + 1):
    if number % i == 0:
        divisors.append(i)
divisors.append(number)
print(divisors)
```

```python
#Q4: Armstrong number
a = 153
copy_a = a
i = len(str(a))
add = 0
while a > 0:
    digit = a % 10
    add = add + (digit**i)
    a = a // 10
if add == copy_a:
    print('Armstrong number')
else:
    print('Not a armstrong Number')
```

```python
#Q5: Prime Number
p = 12
is_prime = True

if p <= 3:
    is_prime = True
else:
    for i in range(2, int(p**0.5) + 1):
        if p % i == 0:
            is_prime = False
            break
if is_prime:
    print('Prime Number')
else:
    print('Not a prime Number')
```

```python
#Q6: Find a GCD of two numbers
n1 = 23
n2 = 24
GCD = 0
for i in range(1, (n2//2) + 1):
    if n1 % i == 0 and n2 % i == 0:
        GCD = i
print(GCD)
```
