```python
value = 121
copy_value = value
#Inititalization
reverse = 0
while copy_value > 0:
    i = copy_value % 10
    reverse = reverse*10 + i
    copy_value = copy_value // 10
print(reverse)
```

```python
if reverse == value:
    print('Palindrome')
else:
    print('Not Palindrome')
```

```python
orignal_balance = 5000
copy_balance = orignal_balance

while True:
    withdraw = int(input('Enter withdrawal amount: '))
    if withdraw == 0:
        print('Thank You')
        break
    print(f'Withdraw: {withdraw}')
    orignal_balance -= withdraw
    print(f'Remaining Balance: {orignal_balance}')
```

```python
x = 58391
is_largest = 0

while x > 0:
    digit = x % 10
    if is_largest <= digit:
        is_largest = digit
    x = x // 10
print(f'Largest digit = {is_largest}')
```

```python
y = 891234567
count = 0

while y > 0:
    is_even = y % 10
    if is_even % 2 == 0:
        count += 1
    y //= 10
print(count)
```

```python
z = 12321
copy_z = z
reverse = 0
while z > 0:
    d = z % 10
    reverse = reverse * 10 + d
    z = z // 10
if reverse == copy_z:
    print('Palindrome')
else:
    print('Not palindrome')
```
