```python
#Q1: Print numbers from 1 to 10.

for i in range(1, 11):
    print(i, end = ' ')
```

```python
#Q2: Print all even numbers between 1 and 100.

for i in range(1, 101):
    if i % 2 == 0:
        print(i, end = ' ')
```

```python
#Q3: Print all odd numbers between 1 and 100.

for i in range(1, 101):
    if i % 2 != 0:
        print(i, end = ' ')
```

```python
#Q4: Print numbers from 10 down to 1

for i in range(10, 0, -1):
    print(i, end = ' ')
```

```python
#Q5: Print the table of a given number (n × 1 to n × 10). 

n = 5

for i in range(1, 11):
    print(n * i, end = ' ')
```

```python
#Q6: Print the sum of first n natural numbers.
num = 10
total = 0

for i in range(1, num+1):
    total += i
print(total)
```

```python
#Q7: Print the sum of all even numbers up to n. 

x = 10
result = 0

for i in range(1, x + 1):
    if i % 2 == 0:
        result += i
print(result)
```

```python
#Q8: Print the sum of all odd numbers up to n.
summition = 0
for i in range(1, x + 1):
    if i % 2 != 0:
        summition += i
print(summition)
```

```python
#Q9: Print the factorial of a given number.
y = 5
fact = 1

while y > 0:
    fact *= y
    y -= 1
print(fact)
```

```python
#Q10:  Print the product of digits of a given number.
number = 123
product = 1

while number > 0:
    digit = number % 10
    product *= digit
    number = number // 10
print(product)
```
