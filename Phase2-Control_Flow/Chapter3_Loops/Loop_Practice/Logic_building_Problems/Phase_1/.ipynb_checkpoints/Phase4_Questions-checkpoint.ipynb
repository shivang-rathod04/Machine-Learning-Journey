```python
#Q1: Print Single star

print('*')
```

```python
#Q2: Print Four starts

print('*'*4)
```

```python
#Q3: print n stars on same line

# Approach 1:
n = 10
print('*'*n)

# Aprroach 2:
for i in range(n+1):
    print('*', end = ' ')
```

```python
#Q4: Print Rectengle 


for i in range(5):
    for j in range(5):
        print('*', end = ' ')
    print()
```

```python
#Q5: Print Increasing Triangles

for i in range(5):
    for j in range(i + 1):
        print('*', end = ' ')
    print()
```

```python
for i in range(5):
    for j in range(5 - i - 1):
        print(' ', end = ' ')
        
    for k in range(i+1):
        print('*', end = ' ')
    print()
```

```python
for i in range(6):
    for j in range(i):
        print('*'*2, end = '')
    print()
```

```python
for i in range(6-1):
    for j in range((i*2) + 1):
        print('*', end = ' ')
    print()
```

```python
for i in range(6):
    # Right angle
    for j in range(6 - i - 1):
        print(' ', end = ' ')
    for k in range((i*2)+1):
        print('*', end = ' ')   
    print()
```

```python
for i in range(8):
    for j in range(i+1):
        if j + 1 < 8 - i:
            print('*', end = ' ')
    print()
```

```python
for i in range(4):
    for j in range(i + 1):
        print('*', end = ' ')
    print()
for j in range(4, 0, -1):
    for k in range(j):
        print('*', end = ' ')
    print()
```
