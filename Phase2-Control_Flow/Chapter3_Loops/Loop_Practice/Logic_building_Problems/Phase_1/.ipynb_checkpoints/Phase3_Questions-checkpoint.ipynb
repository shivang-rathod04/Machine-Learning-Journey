```python
#Q1: Print the squares of numbers from 1 to n.
n = 10
for i in range(1, n+1):
    print(i**2, end = ' ')
```

```python
#Q2: Print cubes of numbers from 1 to n
for i in range(1, n+1):
    print(i**3, end = ' ')
```

```python
#Q3: Print all numbers between a and b divisible by 7.

a = 1
b = 100

for i in range(a, b+1):
    if i % 7 == 0:
        print(i, end = ' ')
```

```python
#Q4: Find HCF (GCD) of two numbers using loops
x = 8
y = 14
prev = 0

for i in range(1, (y // 2) + 1):
    if x % i == 0 and y % i == 0:
            prev = i
print(prev)
```

```python
#Q5: Find LCM of two numbers using loops. 
```

```python
# Approach 1: 
m1 = 9
m2 = 18
l = float('inf')

for i in range(1, 11):
    left = m1 * i
    for j in range(1, 11):
        if left == m2 * j:
            if l >= left:
                l = left
print(l)
```

```python
# Approach 2:
# 1. take the HCF
# 2. Then multiply n1 x n2 
# 3. divide the result // GCF
v1 = 9
v2 = 18
highest = 0

for i in range(1, (v2 // 2) + 1):
    if v1 % i == 0 and v2 % i == 0:
            highest = i
            
LCM = (v1 * v2) // highest
print(LCM)
```

```python
#q6: Print all factors of a given number.
factor = 36
for i in range(1, (factor) + 1):
    if factor % i == 0:
        print(i, end = ' ')
```

```python
#Q7: Find the sum of all factors of a number. 
q = 14
t = 0
for i in range(1, q+1):
    if q % i == 0:
        t += i
print(t)
```

```python
#Q8: Check if a number is a strong number (sum of factorials of digits = number).

number = 145
copy_number = number
total = 0

def factorial(n):
    fact = 1
    while n > 0:
        fact *= n
        n -= 1
    return fact
    
while copy_number > 0:
    digit = copy_number % 10
    total += factorial(digit)
    copy_number = copy_number // 10
if total == number:
    print('Strong Number')
else:
    print('Not a strong number')
```

```python
#Q9: Print first n terms of an arithmetic progression (a, d). 

a = 2
d = 3
n = 6

while n > 0:
    print(a, end = ' ')
    a += d
    n -= 1
```

```python
#Q10: Print first n terms of a geometric progression (a, r).
start = 3
term = 6
r = 2

while term > 0:
    print(start, end = ' ')
    start *= r
    term -= 1
```
