```python
#Q1: Count the number of digits in a given number. 

n = 1236
count = 0

while n > 0:
    n = n // 10
    count += 1
print(count)
```

```python
#Q2: Print the reverse of a given number
x = 123
reverse = 0

while x > 0:
    digit = x % 10
    reverse = reverse * 10 + digit
    x = x // 10
print(reverse)
```

```python
#Q3: Check if a number is a palindrome.
orignal = 121
copy = orignal
r = 0
while copy > 0:
    digit = copy % 10
    r = r * 10 + digit
    copy = copy // 10
if r == orignal:
    print('Palindrome')
else:
    print('Not Palindrome')
```

```python
#Q4: Find the sum of digits of a number. 

y = 123
total = 0

while y > 0:
    digit = y % 10
    total += digit
    y = y // 10
print(total)
```

```python
#Q5: Find the Armstrong number
number = 153
i = len(str(number)) # Alternate : math.floor(math.log10(number)) + 1
result = 0
is_equal = number

while number > 0:
    digit = number % 10
    result = result + (digit**i)
    number = number // 10
if result == is_equal:
    print('Armstrong')
else:
    print('Not Armstrong')
```

```python
#Q6:Check if a number is a perfect number.

z = 8128
summition = 0

for i in range(1, (z // 2 ) + 1):
    if z % i == 0:
        summition += i
if summition == z:
    print('Perfect number')
else:
    print('Not a perfect number')
```

```python
#Q7: Print all prime numbers between 1 and 100. 
primes = []
for i in range(1, 101):
    if i <= 3:
        primes.append(i)
        continue
        
    is_prime = True
    
    for j in range(2, int(i**0.5) + 1):
        if i % j == 0:
            is_prime = False
            break
        
    if is_prime:
        primes.append(i)
print(primes)
            
        
```

```python
#Q8: Check if a number is prime or not.

p = 4
is_prime = True

if p <= 3:
    is_prime = True
else:
    for i in range(2, int(p**0.5) + 1):
        if p % i == 0:
            is_prime = False
            break
if is_prime:
    print('Prime Number')
else:
    print('Not a prime number')
```

```python
#Q9: Print Fibonacci series up to n terms.

f = 8 # Number of terms want to print
a = 0
b = 1
print(a, end = ' ')
print(b, end = ' ')

for i in range(f - 2):
    c = a + b
    print(c, end = ' ')
    a = b
    b = c
```
