```python
# Without Comments

text_data = "My name is shivang and i am from Gujarat. I want to become (Machine Learning Engineer)"
bucket = text_data.split()
name = bucket[3]
print(name)
```

```python
# With comments

# Load the actual data
text_data = "My name is shivang and i am from Gujarat. I want to become (Machine Learning Engineer)"

#Split all the words and create list of words
bucket = text_data.split()

#Extract the name from the data
name = bucket[3]

#Print the name
print(name)
```
