```python
# 3. Escape sequence characters: 
# - In python there are two types of characters : 1. normal charachters and 2. Special Characters
# - normal characters include: a-z, 0-9, symbols
# - special characters include: character that starts with backslash (\) and follows with a character or symbol
```

```python
# What is purpose of escape sequence character ?
# - It allows to insert the characters in strings that is illegal or breaks your code. By inserting the special characters. it tells python to process the string with special meaning.
```

```python
# what do we mean by "meaning"
# - there are 6 types of escape sequence characters in python. each servers different meaning and purpose
```

```python
# 1. \' and \"
# - This is used when you want include the qoutes inside the string.
```

```python
# Eg: Without escape sequence 
print("Hello "world"")

# When we include the quoutes inside string it gives syntax error. Reason is : it thinks that "Hello " is an valid string and the world has no meaning and last "" is an empty string
# we broke the rule of python and hence it returns syntax error
```

```python
# To use the qoutes inside your string There are two ways:
# 1. use escape sequence character
print("Hello \"World\"")
```

```python
#2. use opposite qoutes based on what outside qoutes you have used
```

```python
print('Hello "World"')
# or
print("Hello 'World'")
```

```python
# 2. \n
# - \n indicates python to break the following string and print the rest of string on new line.
# - if no strings follows \n, it automatically print's new line. 
print("message1\nmessage2")
print("message1\n")

# If you don't provide any argument to print. It automatically prints empty line.
print()
# Reason : By default print has set it's end parameters as : \n. so if no string is provided it print's empty line
# you can change the parameter default value and set according to your style
print("hello", end = " ")
print("world")
```

```python
# Instead of typing \n repeatedly. more better approach is to use triple qoutes. it allows to write string in new line and also automatically set's new line
print("""Hi my name is shivang and 
I wanted to tell you that
This is python course""")
```

```python
# 3. \t and \b

# This Tells python provide the following string with tabspace [\t]
print("My name is:\tshivang")
```

```python
# 4. double backslash
# This indicates python that treat this backslash as normal character instead of special character
print("Path: c:\\shivang\\desktop")
```

```python
# you can also use raw string {r""}
# - this is used mostly when work with file path
# As this eliminates the backslash processing
print(r"c:users\desktop\pyhton_course")
```

```python
# The sep parameter in python
# sep parameter is used to seperate the arguments with particular format.
# By default it's a whitespace

print("Hello", "World", sep = "-")
```

```python
# 4. f-strings in python
# - it is modern pythonic way to write formated strings. This allows flexiblity and more readable approach to write strings and perform operation instead of using commma
# - without f-strings it will be very complex to use comma for getting output. and increases chances of syntax error or type error
#Eg : 
print("My name is Shivang and my age is: ", 23, "I have experience of: ", 2, "years and today's date is ", 24)

# Better approach is to use f-strings
print(f"My name is shivang and my age is {24} and I have experience of {2} years and today's date is {24}")
# You can perform any operations inside the curly brackets.
```

```python
# Number formating with f-strings
# - f-strings allows to format a floating point numbers [by rounding it of]
# Eg: rounding of/
num = 123.4567
print(f"{num:.2f}") # syntax: f"{variable:.specifier}"
```

```python
# 2. Applying the underscore or comma to large numbers
n = 100000000000000000000
print(f"{n:_}")
# or
print(f"{n:,}")
```

```python
# 3. Use for calculating percentage for floating point numbers
accuracy = 0.9876
print(f"{accuracy:.1%}")
```

```python
# You can perform multiple modifiers together
# Eg:
price = 1899232233.890956
print(f"House Price = {price:,.2f}")
```

```python
# Placing = after both variable and expression gives the current state/value of variable and expression.
# It's allowed in python from Python 3.8

x = 20
y = 89
print(f"{x = } {y = } {x + y = }")
```
