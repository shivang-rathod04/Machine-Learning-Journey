```python
# Hardcoded values vs Dynamic values
# Hardcoded values: are fixed, they need explicit modification to change the state
# Dynamic values : are not fix, they are created and assigned during execution time.

# Hardcoded value 
def greet(name):
    location = "Gujarat" # Hardcoded value
    return f"Welcome {name} you are from {location}"

name = input("Enter your name: ") # dynamic value

greet(name)
```
