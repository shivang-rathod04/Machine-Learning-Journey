```python
# Python handles the foating point numbers - using IEEE 754 - Double precision floating point standard.
# This is a blueprint that is used to store the numbers in binary form
```

```python
# In python every float value follows this standard under the hood, also called float64 [in languages like c/c++ or numpy]
```

```python
# How This standard works inside the memory 
```

```python
# Inside the memory, every floating value is been allocated only 64 bits [8 bytes] for every floating number
```

```python
# Now this 64-bits are divided into 3 parts : 
# 1. sign bit - indicates wheather number is positive or not 
# - if number is positive [0] and if number is negative [1]
# - It takes only one bit
```

```python
# 2. exponent - tells how big the number is or how small the number is.
# - It scales down the number by power of 2
# - It takes 11 bits
```

```python
# 3. Mentasia - This are the significant bits
# - It stores actual precision digits.
# - Takes around 52 bits.
```

```python
# From this there are three rules: 
```

```python
# 1. Binary Fraction Problem
# - Not all binary fractions are fixed interpreations
# - some numbers like 0.1, binary representation is infinite 
# as only 52 bits are available, infinite bits cannot be stored inside memory
# hence, It cuts off the rest of the bits
# Due to this, there's always precision error. 
```

```python
result = 0.1 + 0.2
print(result == 0.3) # This will return False
```

```python
# 2. Precision Problem
# - IEEE 754 standard can only store about 15-17 significant digits precisely, rest of the bits are truncted or rounded
# - Due to this there's always precision error inside floating point numbers
```

```python
# Hence, to compare floating point numbers you must always you .isclose() method
# - isclose makes sure weahter two of the numbers are close enough. if yes than it returns true
```

```python
# .isclose method belongs to math module. 
# so we need to import in-order to use this function
```

```python
import math

print(math.isclose(result, 0.3))
```

```python
# You can even set the tolrence [for normal tolrence you can use 10^9]
print(math.isclose(result, 0.3, rel_tol = 1e-9))
```

```python
# For absolute tolrence you use [10^-3]
print(math.isclose(result, 0.3, rel_tol = 10e-3))
```

```python
# How do we know which number binary interpration is infinite or finite
# 1. Float - whole numbers are exact in binary
# 2. Float - powers of 2 are exact inside memory. [0.5, 0.25, 0.125, etc]
# 3. Float - fractions like 0.1, 0.3, 0.4, 0.6 etc.. binary representaion is infinite
```

```python
# NaN [Not a Numberp]
# It occurs when standard rules of operations are break [invalid operations] or when a value is not defined
```

```python
# How to create a nan in python
```

```python
nan_val = float("nan")
print(type(nan_val))
print(nan_val)
```

```python
# Rule you must remember:
# - NaN can never be equal to any object, not even with itself!!

print(nan_val == nan_val) # this will return "False"
print(nan_val != nan_val) # this will return "True"
```

```python
# any operation done on nan will always return nan
addition = nan_val + positive_inf
print(addition)

# adding positive infinity and negative infinity will result nan
print(positive_inf + negative_inf)

# division of invalid operation will yeild a nan
print(positive_inf / negative_inf)
```

```python
# Reason why we need to remember This rule is 
# - when we are dealing with EDA [Exploratery data analysis]. process when manuplating the data in Machine Learning
# - if any  value [feature] is 'nan' then overall result will be nan
# - this will affect while you are training your model.
```

```python
# Complex numbers are third catagory of numbers
# This are also considered as primitive types
# Complex numbers are divided in two parts : real and imaginary 
# This are also an immutable types - changing them would create new object
```

```python
# To create a complex number
# There are two ways:
version1 = 3 + 4j
version_2 = complex(2, 3)

print(version1)
print(version_2)
```

```python
# You won't use the complex number much for your day to day programming
# as complex number are used for advanace methamatical operations or for scientific calculation
# signal processing, fourier transforms [used in audio ml]
```

```python
# some of the operations and method for complex numbers
```

```python
print(version1)
print(version1.real) # returns real part of the number
print(version1.imag) # returns imaginary part of number
print(version1.conjugate())
print(abs(version1))

# operation on complex number
print(version1 * version_2)
print(version1 + version_2)
```
