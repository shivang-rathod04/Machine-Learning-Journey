```python
# Now this data types are divided into two catagories:
# Mutable types and Immutable types
```

```python
# Immutable types : 
# Immutable types are those which cannot be modified [in-place after the creation]
# Eg: int, float, string, tuples, boolean
# what does it mean ?
```

```python
# when you try to perform any modification on the immutable types. python does not modify that orignal data instead it creates new object and 
# bounds the variable to that new object. It's concept of "rebounding"
```

```python
a = 10
b = a 
a = 20

print(f"Value of a: {a=} | Value of b: {b=} ")
```

```python
# when you write : 
# a = 10 : python creates an object of type int and bounds variable a to that object
# in memory a -> 10 [ref count = 1]
# b = a : python doesn't create a new object. it bounds the variable b to the same object where a is pointing towards
# therefore in memory : a -> 10 <- b [ref count : 2]

# now just see that we used integers !!.. and we know integers are immutable. means python will create new object instead of modifying the orignal
# therfore :
# on a = 20 : python creates new object inside memory and "rebounds" the variable a to the new object
# in memory : b -> 10 [ref count becomes 1] and a -> 20 [ref count : 1]
# thefore final answer becomes : a = 20 and b = 10
```

```python
# If no reference [variable] is pointing towards the object inside memory, it becomes unreachable and becomes eligible for garbage collection.
```

```python
# From this always remember immtables types are always rebounded on performing the operations or try to modify them
```

```python
# 2. Mutable types:
# - mutable objects can be modifed in-place. that means unlike immutable, it will modify the orignal data instead of creating new object. 
# By this changing the orignal data might also affect all the variables that pointing towards that object.
# Now, weather the perticular operation can be modified in-place or variable rebounds depends on what kind of operation is performed
```

```python
#Eg:
data = [1, 2, 3]
copy_of_data = data
print(f"Orignal Data Before : {data = } | Copy data before : {copy_of_data = }")

# Appemding the new element in the orignal data.
data.append(4)

print(f"Orignal Data after: {data = } | Copy of data after: {copy_of_data = }")
```

```python
# We didn't touched copy of the data. still it modified the copy of data. how ?
# process is same:
# data = [1, 2, 3] : python creates an object of type list. which holds the references of elements inside memory
# in memory : data -> [list object]
#                      |    |    |
#                      1    2    3
# when we create copy of data = data 
# python bounds the copy_of_data to the list object where data is pointing towards
# therfore in memory : 
# # data -> [list object] < - copy_of_data
#            |    |    |
#            1    2    3

# now python sees "what operation is performed" and what type of data it is.
# python sees type of object is list which is mutable object. and here .append() is performed. [.append() modifes the object in place and affects all the variables pointing towards this object.]
# so chaning the data also changed the copy_of_data.
```

```python
# There are some operations - for mutable which doesn't perform in-place modification. instead it creates new object and variable re-bounds

new_data = [5, 6, 7]
new_copy = new_data
print(f"State of new_data before modifiying : {new_data = } | State of new_copy before modifying : {new_copy = }")

# here we are adding two list.
new_data = new_data + [3]

print(f"State of new_data after modifiying : {new_data = } | State of new_cop after modifying : {new_copy = }")
```

```python
# Here adding of two list. made new_data rebound to the new list object inside memory. and copy of data is untouched
```

```python
# From this we understand that in python mutable objects can be modified in place and also there can be concept of variable rebounding for some operations.
```

```python
# How would we know which operation creates new object and which modifies in place for mutbale objects ?
# Simple: 
# mutable object methods always modifes in place
# sorted(), slicing, adding two mutbale objects. creates new object inside memory and performs rebounding
```
