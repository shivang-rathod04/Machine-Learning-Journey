```python
# Understanding booleans inside python
```

```python
# Booleans in python are divided into two catagories :
# True 
# False
```

```python
# How to create a booleans 
x = True
y = False
z = bool(0) # Returns False for empty objects and True for non-empty objects

print(type(x))
print(type(y))
print(type(z))
```

```python
# Booleans are also the singleton objects like None
# Means there's only one True object - no matter how many you create all references will point to the same obj

value_1 = True
value_2 = True
value_3 = True

print(value_1 is value_2)
print(value_2 is value_3)
```

```python
# Booleans doesn't have independent type like other objects, 
# Under the hood booleans are sub-class of integers.
# Therefore, True - 1 and False - 0
```

```python
print(isinstance(True, int))
print(isinstance(False,int))
```

```python
print(True == 0)
print(True == 1)
print(False == 0)
print(False == 1)
```

```python
# You can perform any mathametical operations on boolean
```

```python
print(True + True)
print(True - 100)
print(True * True)
print(False ** True)
print(True / True)
print(True // True)
print(True % True)
```

```python
# In python Every object can be evaluated in booleans [either true or false]
# This is concept of Truthiness
# where some objects inside python always remain false. understanding them is crucial
```

```python
# The rule is simple
# 0, 0.0,0j, None, "", {}, [], (), set() [This are all falsy values]
```

```python
# All falsy values inside python

print(bool(0))
print(bool(0.0))
print(bool(0j))
print(bool(None))
print(bool({}))
print(bool([]))
print(bool(set()))
print(bool(()))
```

```python
# Some of the values that seem falsy but they are true values
```

```python
lst = [[[]], {"":""}, " ", [""]]

for items in lst:
    if items:
        print(f'True values : {items}')
    else:
        print(f'False values : {items}')
```

```python
# some of the values that seem true but that are false

print(bool((0)))
print(bool(set([])))
```

```python
# Boolean operators
# and - Returns true only when both values are true
# or - Returns true if any one the value is true 
# not - reverses the value [if true - returns as false, same vice versa]
```

```python
# and
print(True and True)
print(True and False)
print(False and False)

# or 
print(True or True)
print(True or False)
print(False or False)

# not

print(not True)
print(not False)
```

```python
# Short circuit operations in python
# - for 'and' if left is false, don't check right
# - for 'or' if left is True, don't check right
```

```python
# and 
print(90 and 100) # returns 100
print('' and {}) # ''
print(3 and {}) # returns {}
print({} and 34) # returns {}
```

```python
# or
print(89 or 65) # returns 89
print('' or {}) # returns {}
print('' or 90) # returns 90
print(34 or '') # returns 34
```

```python
def is_valid_probability(n):

    if not isinstance(n, (int, float)):
        return False
    else:
        return 0 <= n <= 1

print(is_valid_probability(0.5))          # True
print(is_valid_probability(0))             # True
print(is_valid_probability(1))             # True
print(is_valid_probability(1.5))          # False — above 1
print(is_valid_probability(-0.1))         # False — below 0
print(is_valid_probability(float('nan'))) # False — NaN
print(is_valid_probability(float('inf'))) # False — infinity
print(is_valid_probability("0.5"))  
```
