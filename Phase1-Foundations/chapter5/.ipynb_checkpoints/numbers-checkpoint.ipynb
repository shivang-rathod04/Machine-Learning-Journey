```python
# Integers are whole numbers without decimal points
# numbers include in integers are : positive, 0, negative
# Integers are immutable in nature : it cannot be modified in-place 
# as a result python always creates new integer and rebounds the variable to new int object
```

```python
positive_int = 23
zero = 0
negative_int = -45

print(type(positive_int)) # All three belong to the 'int' class
print(type(zero)) # All three belong to the 'int' class
print(type(negative_int)) # All three belong to the 'int' class
```

```python
# Python handles integers differently as compared to other languages:
```

```python
# Other languages like c or java, uses numbers in fixed bit [int32 or int64]
# If we consider int32, maximum number that can represent in int32 is 2.1 X 10^9
# If any number goes above that range, it causes "integer overflow"
# and number wraps around negative number.
# It doesn't mean answer doesn't exist. 
# Solution exists but there are no bits left to represent that number inside memory. as all bit patterns are used to represent number
# That's why it causes integer overflow
```

```python
# But in python, It doesn't have any fix size
# number can be any large.
# As the number grows, python allocates more memory to it.
# It's called Arbitary Precision Integers in python
```

```python
large_number = 10000000000000000000000000000000000
print(f"Large number : {large_number:,}")
```

```python
# But it comes up with cost. if number increases, python allocates more memory.
# and this can cause performance issue. and heavy memory usage
```

```python
# But Third-party library of python are built under c/c++ and it heavily uses concept of integer overflow and at that time we need to consider this before using integers
# That's why i call python integers as "Sweet but poisonous"
```

```python
# Concept of Type Conversion for integers:
# Python allows explicit type conversions from one data type to another.
# For integers you can convert {int -> float, int -> complex, int -> string}
# let's try
```

```python
int_to_float = float(3)
int_to_complex = complex(2, 3) # as we know complex has two parts : real and imaginary [here 2 is real | and 3 is imaginary]
int_to_string = str(3)

print(type(int_to_float))
print(type(int_to_complex))
print(type(int_to_string))
```

```python
# You can perform all mathametical operations on integers

print(2 + 3) # performs addition
print(2 - 3) # performs substraction
print(2 ** 3) # It works as 2^3. exponent.
print(2 * 3) # performs multiplication
print(2 % 3) # Gives remainder [either 0 or 1]. if number divides perfectly it returns 0 and if it's not divisible it returns 1
print(2 / 3) # performs True division [ always returns output in floating point number, even if it divides perfectly]
print(2 // 3) # perform floor division [ rounds number to nearest smallest integer. [we'll dive deep in concept of rounding numbers]
```

```python
# Let's Understand how python stores integers inside memory
# We know everything inside python is an object
# Inside memory python doesn't only store the value. it stores object + metadata
# Meta data includes extra information like : 
# - ref_count : number of variables pointing towards this object
# - type_of_object : what type of object it is [int, float, etc]
# - actual_value : number. whose size is 4 byte
# the metadata takes up extra space. 4 byte of integer becomes of 28 byte
```

```python
import sys

get_int_size = 44
print(f"Size of integers in python: {sys.getsizeof(get_int_size)}")
```

```python
# But again using third-party library removes all the extra information. what it stores is value with no extra information
```

```python
# you can also convert decimal to hexadecimal, octal, binary

print(bin(15))
print(oct(13))
print(hex(22))
```

```python
# Concept of integer catching:
# Python has concept of integer caching, 
# where python pre-creates small integers [-5 to 256] when script is loaded
# so when object is created in that range, python re-uses that same cached int inside memory. instead of creating new object
```

```python
small_number = 13
large_number = 257
print(id(small_number))
print(id(large_number))
```

```python
# You can see that id of small number remains same, where id of large number always change
# is because for small number python re-uses same object that's pre-created
```

```python
# point of learning this is - 
# when we compare small integers. we cannot use "is" as it will return True
# because 'is' checks if both the objects are same inside memory or not. if it's true it returns true or false
# so we always use == to comapre the numbers. instead of using is
# This is implementaion detail. for different python version. implementaion may vary.
```
