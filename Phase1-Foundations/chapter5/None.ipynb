```python
# Understanding NONE object inside python
```

```python
# 'None' is python's way to represent 'nothing'/'no value' etc.
```

```python
# To create none object 

none_obj = None
print(none_obj)
```

```python
# None object has it's own type [NONE belongs to class NoneType]
# That is - None Type
```

```python
print(type(none_obj))
```

```python
# In python, There's only one None object
# No matter how much None object you create, every references will point to that same none object inside memory
# That means, python pre-creates this None object inside memory.
# so every time when you try to create new none object, referemces point to the same pre-created none object inside memory
# instead of creating new one
```

```python
print(id(none_obj))
```

```python
new_obj = None
print(id(new_obj))
print(new_obj is none_obj) # This will return TRUE 
```

```python
# A None is an Singleton Object
# There are few more singleton objects inside python 
# Eg: True, False [this we will study in boolean]
```

```python
# The common use cases of none object inside python
```

```python
# 1. Sentinetal value
# - this is used when you don't have initital state/baseline before performing operation
# - you set the value as None for placeholder, so the variable exists and points to the reserved space inside memory
goal = None
print(f"Count before : {goal}")
for i in range(3):
    if i == 2:
        goal = "Yay"
        break
print(f"Count after : {goal}")
```

```python
# 2. using None as default arguments
```

```python
# - using none as default argument is prefered because during function defination type, mutable object are only
# created once, so even if you call function multiple time it will re-use same mutable of object instead of creating new one
# therefore for mutable object default argument is set as none 
```

```python
def normal(data, arr= None):
    if arr is None:
        arr = []
    return data
```

```python
# 3 sending signal that function found nothing

def finding(num):
    for n in num:
        if n < 0:
            return n
    return None

print(finding([1, 2, 3]))
print(finding([1, -2, 3]))
```
