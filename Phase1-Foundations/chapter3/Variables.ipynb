```python
# Understanding variables
```

```python
# Variables are the lables/references/pointer that are bound towards the object inside memory
# Object means [integer, float, string etc] even an function is an object in python. Everything inside python is an object
# Object lives independently inside memory when created, variables are just way to trace those object [Access and reading]
```

```python
# What do we mean by object ?
# Object means - any data piece you see in python is an object
# each object has a fixed properties
```

```python
# 1. Memory address : Location where object is living/stored
# 2. reference count : How many variable are pointing to this object
# 3. Type: What type of object it is [integer, float, string, boolean etc]
# 4. Actual data [state]
```

```python
# How to declare and initialize variable:
# variable_name = value

number = 45
print(f"Number is: {number}")
```

```python
# Can we access the properties of this object
# Getting id [memory address of object]
print(id(number))
```

```python
# Getting Type() of this object
print(type(number)) # returns class of int
```

```python
# Reference count of object 45 is only 1, because only one variable is pointing towards that object inside memory.
```

```python
# actual data
print(number)
```

```python
# Naming conventions for variables
# - Variables cannot start with numbers
# - variables not contains whitespace or tabspace
# - variables must not contains symbols except underscore.
# - variable names must have meaning
# - keywords cannot be used as variable name
# - Python is case-sensetive, i.e {name} amd {Name} are two different meanings for python
```

```python
# Variables styles according to [pep 8 writing style guide]
```

```python
# PascalCase: 
# - starting letter of all words starts with capital letters
# - does not contain underscore
# - This style is used when you are working classes. classes name are written pascalCase
# Eg:
MyClass = "GetPrimeNumber"
print(MyClass)
```

```python
# SnakeCase:
# - all the words are in lower letters
# - can contain underscores
# - this is used when you are declaring variable names or function names

def my_function(name):
    return f"Hello {name}"

my_name = "Shivang"
my_function(my_name)
```

```python
# Assigning multiple values to multiple varaibles
x, y, z = 1, 2, 3
print(f"{x=}, {y=}, {z=}")
```

```python
# Assinging same value to multiple variables
x2 = y2 = z2 = 6
print(f"{x2 = }, {y2 = }, {z2 = }")
```
