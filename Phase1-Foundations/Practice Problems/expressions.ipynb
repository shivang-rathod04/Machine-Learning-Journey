```python
# Write a program to obtain x, y and z from user and calculate expression: 4x4 + 3y3 + 9z+6π
```

```python
import math as mp
x = float(input("Enter value of x: "))
y = float(input('Enter value of y: '))
z = float(input('Enter value of z: '))
```

```python
result = (4 * (x**4)) + (3 * (y**3)) + (9 * z) + (6 * mp.pi)
print(f'Result = {result:.2f}')
```
