```python
# Write a program to read base/length, width and height of a parallelogram 
# and calculate its area and perimeter.
```

```python
base = float(input('Enter a base of parallelogram: '))
width = float(input('Enter a width of parallelogram: '))
height = float(input('Enter a height of parallelogram: '))

area = base * height
perimeter = 2 * (base + width)

print(f'Area of Parallelogram: {area:.2f}')
print(f'Perimeter of Parallelogram: {perimeter:.2f}')
```
