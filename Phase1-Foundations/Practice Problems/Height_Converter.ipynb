```python
# Write a Python program that asks for your height in centimetres and then converts your height to feet and inches. (1 foot = 12 inches, 1 inch =  2.54 cm).
```

```python
user_height = 240

# converting cm to inch
inch_height = user_height / 2.54

# converting inch to foot
foot_height = inch_height / 12

print(f'Height is {foot_height:.2f} foot')
```
