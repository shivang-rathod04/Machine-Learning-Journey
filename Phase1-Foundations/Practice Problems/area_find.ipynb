```python
# Find area of triangle, square, rectengle, circle
```

```python
# Triangle
base = 3
height = 5

area_triangle = 0.5 * (base * height)
print(area_triangle)
```

```python
# square
side = 3

area_square = side * side
print(area_square)
```

```python
# circle
import math as mp

radius = 3

area_circle = mp.pi * (radius**2)
print(round(area_circle, 2))
```

```python
# rectengle
length = 2
width = 2

area_rectengle = length * width
print(area_rectengle)
```
