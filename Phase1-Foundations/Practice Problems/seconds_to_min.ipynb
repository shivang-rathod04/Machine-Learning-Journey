```python
# Write a program that reads a number of seconds and prints it in form: min and seconds, e.g.,
# 200 seconds are printed as 3 mins and 20 seconds.
```

```python
# Method 1:
time = 254

# converting seconds to minute
minute = time // 60

# Getting seconds [formula : Orignal seconds - 60 * qoutient]
seconds = time - (60 * minute)

print(f'{minute} min : {seconds} sec')
```

```python
# Method 2:

user_input = 500

result = divmod(user_input, 60) # divmod(dividend, divisor) = (qoutient, remainder)
print(f'{result[0]} min : {result[1]} sec')
```
